```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set(style="whitegrid")
%matplotlib inline

# 📂 Load the dataset
train_df = pd.read_csv("train.csv")

# 👀 Preview the data
train_df.head()
```

1. Basic Info & Summary

```python
# Structure and missing values
train_df.info()
```

```python
# Summary statistics
train_df.describe(include="all")
```

2. Univariate Analysis

Survival Count

```python
sns.countplot(data=train_df, x='Survived')
plt.title("Survival Count")
plt.show()
```

Passenger Class

```python
sns.countplot(data=train_df, x='Pclass')
plt.title("Passenger Class Distribution")
plt.show()
```

 Sex Distribution

```python
sns.countplot(data=train_df, x='Sex')
plt.title("Gender Distribution")
plt.show()
```

Embarkation Port

```python
sns.countplot(data=train_df, x='Embarked')
plt.title("Port of Embarkation")
plt.show()
```

Age Distribution

```python
sns.histplot(train_df['Age'].dropna(), kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()
```

Fare Distribution

```python
sns.histplot(train_df['Fare'], kde=True)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.show()
```

3. Bivariate Analysis

 Survival by Sex

```python
sns.countplot(data=train_df, x='Sex', hue='Survived')
plt.title("Survival by Sex")
plt.show()
```

 Survival by Class

```python
sns.countplot(data=train_df, x='Pclass', hue='Survived')
plt.title("Survival by Passenger Class")
plt.show()
```

Age vs Survival

```python
sns.boxplot(data=train_df, x='Survived', y='Age')
plt.title("Age Distribution by Survival")
plt.show()
```

Fare vs Survival

```python
sns.boxplot(data=train_df, x='Survived', y='Fare')
plt.title("Fare Distribution by Survival")
plt.show()
```

Correlation Matrix

```python
corr = train_df.corr(numeric_only=True)
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

5. Pairplot

```python
sns.pairplot(train_df[['Survived', 'Pclass', 'Age', 'Fare']].dropna(), hue='Survived')
plt.suptitle("Pairwise Relationships", y=1.02)
plt.show()
```

6. Observations & Insights

- Observation 1: Majority of survivors were in 1st and 2nd class.
- Observation 2: Women had higher survival rates than men.
- Observation 3: Younger passengers slightly more likely to survive.
- Observation 4: Fare positively correlated with survival.
